```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

df = pd.read_csv('test.csv')
df.head()
```

```python
y = df["Aggregate rating"]
X = df.drop(columns=["Aggregate rating"])

print("Target: Aggregate rating")
print("X shape:", X.shape, "| y shape:", y.shape)
```

```python
class_counts = y.value_counts()

valid_classes = class_counts[class_counts >= 2].index

mask = y.isin(valid_classes)

X = X[mask]
y = y[mask]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print("Train shape:", X_train.shape)
print("Test shape :", X_test.shape)
```

```python
print(y.head())
print(y.dtype)
print(y.nunique())
print(y.value_counts())
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

# Linear Regression
linear_reg = LinearRegression()
linear_reg.fit(X_train, y_train)

# Decision Tree
dt = DecisionTreeRegressor(random_state=42)
dt.fit(X_train, y_train)

# Random Forest
rand_forest = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
rand_forest.fit(X_train, y_train)

# XGBoost
xgb = XGBRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=6,
    random_state=42
)
xgb.fit(X_train, y_train)

print("All regression models trained successfully!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def evaluate_regression(name, model, X_test, y_test):
    preds = model.predict(X_test)

    mae = mean_absolute_error(y_test, preds)
    mse = mean_squared_error(y_test, preds)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, preds)

    print(f"\n{name}")
    print(f"MAE  : {mae:.4f}")
    print(f"MSE  : {mse:.4f}")
    print(f"RMSE : {rmse:.4f}")
    print(f"R²   : {r2:.4f}")

    return preds
```

```python
linear_preds = evaluate_regression(
    "Linear Regression", linear_reg, X_test, y_test
)

dt_preds = evaluate_regression(
    "Decision Tree", dt, X_test, y_test
)

rf_preds = evaluate_regression(
    "Random Forest", rand_forest, X_test, y_test
)

xgb_preds = evaluate_regression(
    "XGBoost", xgb, X_test, y_test
)
```
